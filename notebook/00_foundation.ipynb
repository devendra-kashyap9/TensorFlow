### Definition

* Scalar: a single number
* Vector: a number with direction (e.g. wind speed and direction)
* Matrix: a 2-d array of numbers
* Tensor: an n-dimensional array of numbers (where n can be any number,
                                             where 0-dimensional is scalar and 1-dimensional is a vector)

> **Creating Random Tesnors**

Random Tensors are Tensors of some arbitary size with random numbers (weights).

````python
## creating random tensors
random_1 = tf.random.Generator.from_seed(42) # set seed for reproducibility
random_1 = random_1.normal(shape=(3, 2)) # create a random tensor with normal distribution
random_2 = tf.random.Generator.from_seed(42)
random_2 = random_2.normal(shape=(3, 2))
````

````python
# Are they equal
random_1, random_2, random_1 == random_2
````

> **Shuffling the Order of elements in Tensors**

````python
## shuffling is neccessary for when you want to shuffle data so the order of the data doesn't effect the learning of the model
not_shuffled = tf.constant([[10, 7],
                            [3, 4],
                            [2, 5]])
````

````python
not_shuffled
````

````python
## shuffle the order of the elements in not_shuffled
## global and operation level random seed
tf.random.set_seed(42) # global level random seed
shuffled = tf.random.shuffle(not_shuffled, seed=42)  # shuffled along the first dimension (rows)
                                                     # operation level random seed
shuffled
````

### Aggregating tensors
> **Aggregating tensor -> condensing them from multilple values down to a smaller values**

* Min
* Max
* Mean 
* Sum and more ...

````python
## create a ransom tensor
E = tf.constant(np.random.randint(0, 100, size=50))
E
````

````python
tf.size(E), E.shape, E.ndim
````

````python
# find the minimum
tf.reduce_min(E), np.min(E)
````

````python
# find the maximum
tf.reduce_max(E), np.max(E)
````

````python
# find the meqan
tf.reduce_mean(E)
````

````python
# sum of tensor
tf.reduce_sum(E)
````

````python
## standard devidation and variance of tesnor
''' 
    The issue is that tf.math.reduce_variance and tf.math.reduce_std require the input 
    tensor to be a floating-point (decimal) or complex type. TensorFlow doesn't automatically 
    "cast" your integers to floats for these specific operations because integer division and 
    variance math can get messy.
'''
E = tf.cast(E, dtype=float)
print('Standard deviation of the tensor: ', tf.math.reduce_variance(E))
print('Variance of the tensor: ', tf.math.reduce_std(E))
````

````python
## Alternarive for variance
import tensorflow_probability as tfp
E = tf.constant(np.random.randint(0, 100, size=50))
E
````

````python
tfp.stats.variance(E)
````
