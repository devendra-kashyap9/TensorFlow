```python
print('Question 1')
print('-------------------------------------------------')

import requests
from bs4 import BeautifulSoup
import nltk
import string

# Fetch Wikipedia page for NASA
url = "https://en.wikipedia.org/wiki/NASA"
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')
text = soup.get_text()

# Task 1: POS Tagging
tokens = nltk.word_tokenize(text)
pos_tags = nltk.pos_tag(tokens)

# Task 2: Remove punctuation symbols
no_punctuation = [word for word in tokens if word not in string.punctuation]

# Task 3: Frequency of all words
freq_dist = nltk.FreqDist(no_punctuation)
word_frequencies = freq_dist.most_common()

# Task 4: Morphological analysis of nouns
nouns = [word for word, tag in pos_tags if tag.startswith('NN')]
morphological_analysis = {noun: len(noun) for noun in nouns}  # Example: Length as a simple morphological feature

# Task 5: Top three words
top_three_words = freq_dist.most_common(3)

# Results
print("1. POS Tagging (first 10 tokens):", pos_tags[:10])
print(f"2. Total words after removing punctuation: {len(no_punctuation)}")
print(f"3. Word frequencies (top 5): {word_frequencies[:5]}")
print(f"4. Morphological analysis of nouns (length): {morphological_analysis}")
print(f"5. Top three words: {top_three_words}")


print('Question 2')
print('--------------------------------------------------------')
# Fetch Wikipedia page for Pune
url = "https://en.wikipedia.org/wiki/Pune"
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')
text = soup.get_text()

# Tokenize and POS tagging
tokens = nltk.word_tokenize(text)
pos_tags = nltk.pos_tag(tokens)

# Extract adjectives
adjectives = [word for word, tag in pos_tags if tag.startswith('JJ')]

# Results
# Print first 20 adjectives
print(f"Adjectives in the text: {adjectives[:20]}")  
```
